```python
import numpy as np
```

```python
import matplotlib.pyplot as plt
import pandas as pd
```

```python
np.random.seed(123)
tails = [0]
for x in range(10):
  coin = np.random.randint(0, 2)
  tails.append(tails[x] + coin)

print(tails)
```

```python
np.random.seed(123)
final_tails = []
for x in range(10000):
  tails = [0]
  for x in range(10):
    coin = np.random.randint(0, 2)
    tails.append(tails[x] + coin)
  final_tails.append(tails[-1])
print(final_tails)
```

```python
plt.hist(final_tails, bins = 10)
plt.show()
```
