# **Customization the plot**

```python
# add more data to year and pop

year = [1800, 1850, 1900] + year
pop = [1.0, 1.262, 1.650] + pop
```

```python
plt.plot(year, pop)
plt.xlabel('Year')
plt.ylabel('Populaton')
plt.title('Year vs Population Projection')
plt.yticks([0, 2, 4, 6, 8, 10],
           ['0', '2B', '4B', '6B', '8B', '10B'])

plt.scatter(year, pop)
plt.show()
plt.clf()
```
