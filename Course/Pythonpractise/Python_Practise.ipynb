# ***Calculator***

```python
num1 = int(input("Enter First Number: "))
num2 = int(input("ENter Second Number: "))

total = num1 + num2
sub = num1 - num2
mult = num1 * num2
division = num1 / num2
remainder = num1 % num2

print("Sum Is: ", total)
print("Subtract is: ", sub)
print("Multiplication is: ", mult)
print("Division is: ", division)
print("Remainder is: ", remainder)
print(f"The Sum is: {total}")
```
