## Which NYC schools have the best math results?

```python
eighty_percent = (80/100) * 800

print(eighty_percent)
```

```python
best_math_schools = schools[schools["average_math"] >= eighty_percent][["school_name", "average_math"]].sort_values(by="average_math", ascending=False)

best_math_schools
```

```python
schools.head(1)
```

```python
schools_1 = schools

schools_1.head()
```

```python
schools_1["total_SAT"] = schools_1[["average_math", "average_reading", 
                                    "average_writing"]].sum(axis=1)

schools_1.head(2)
```

```python
top_10_schools = schools_1[["school_name", "total_SAT"]].sort_values(by="total_SAT", ascending=False).head(10)

top_10_schools
```

```python
schools.head(2)
```

```python
borough_states = schools_1.groupby(["borough"]).agg(num_schools=("school_name", "count"), average_SAT=("total_SAT", "mean"), std_SAT=("total_SAT", "std")).round(2)
borough_states
```

```python
largest_std_dev = borough_states.sort_values(by="std_SAT", ascending=False).reset_index().head(1)

largest_std_dev
```
